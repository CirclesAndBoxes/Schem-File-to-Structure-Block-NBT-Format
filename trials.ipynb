```python
from PIL import Image
import os

targeted_folder = r"C:/Users/Public/PrismLauncher/instances/26.1.2/.minecraft/saves/Cell Museum (May 18, 2026) Editing/datapacks/box/assets/celledit/textures/painting/martini3"
MAX_BYTES = 1024 * 1024  # 1 MB

def resize_to_limit(path, max_bytes=MAX_BYTES):
    file_size = os.path.getsize(path)
    if file_size <= max_bytes:
        return

    img = Image.open(path)
    w, h = img.size

    while file_size > max_bytes:
        w, h = max(1, int(w * 0.5)), max(1, int(h * 0.5))
        img = img.resize((w, h), Image.LANCZOS)
        img.save(path, optimize=True)
        file_size = os.path.getsize(path)

    print(f"Resized: {path} ({file_size / (1024 * 1024):.2f} MB)")

for dirpath, _, filenames in os.walk(targeted_folder):
    for filename in filenames:
        if filename.lower().endswith((".png", ".jpg", ".jpeg", ".webp")):
            resize_to_limit(os.path.join(dirpath, filename))
```
