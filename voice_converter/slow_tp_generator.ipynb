### Imports

```python
import os
```

```python
# Voice File Length: ryan[i] length in seconds
voice_lengths = [5, 4, 5, 8, 6]

# Locations: x, y, z, yaw, pitch in Minecraft.
location_text = """
1 2 3 10 11
2 3 4 11 12
3 4 5 12 13
"""
location_array = [list(map(float, line.split())) for line in location_text.strip().split("\n")]
print(location_array)
```

### Data

```python
try:
    os.mkdir("slow_tick")
except:
    Exception

tick_file = open("slow_tick/slow_tick.mcfunction", "w")
```

```python
# Header
tick_file.write("execute if score @s ce.camera_tick matches 50 run data merge entity @s {teleport_duration:0}\n")
```

```python
# Each command:
tick_value = 60

for i in range(len(voice_lengths)):

    tick_file.write(f"execute if score @s ce.camera_tick matches {tick_value} run particle large_smoke ~ ~-0.2 ~ 0.05 0.05 0.05 0.01 25 force\n")
    tick_value += 10
    tick_file.write(f"execute if score @s ce.camera_tick matches {tick_value} run function celledit:other/cinema/functions/blind_tp {{x: {location_array[i][0]}, y: {location_array[i][1]}, z: {location_array[i][2]}, rot_x: {location_array[i][3]}, rot_y: {location_array[i][4]}}}\n")
    tick_value += 10
    tick_file.write(f"execute if score @s ce.camera_tick matches {tick_value} as @p at @s run playsound celledit:martini2_narration_{i} master @s ~ ~ ~ 1000 1 1\n")

    tick_value += voice_lengths[i] * 20 + 10
    
```
